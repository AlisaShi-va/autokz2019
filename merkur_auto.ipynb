## Введение

Исходный массив содержал 39965 записей, содержащий информацию о сделках по продаже транспортных средств, включая технические характеристики автомобилей. Первичный аудит выявил критический уровень технологического шума, дублирование сущностей, аномалии в показателях и другие неточности. В рамках исследования была выполнена сквозная очистка, восстановление логических связей в данных, анализ ситуации на рынке и финальная выгрузка структурированного массива для бизнес-анализа

*Цель работы* - преобразование сырого датасета в очищенную, структурированную базу данных для позиционирования и определения эффективности компании на рынке

## Актуальность работы
В условиях высокой конкуренции на автомобильном рынке Казахстана принятие управленческих решений на основе неточных, «грязных» данных ведет к прямым финансовым потерям и другим возможным рискам. Искажения в отчетности маскируют реальные тренды и приводят к неэффективному планированию складских запасов, неверному распределению бюджетов и упущенной выгоде в высокомаржинальных сегментах, таких как автокредитование и Trade-In. Чистые данные позволят холдингу находить истинные точки роста, оптимизировать цепочки поставок и точно оценивать свои рыночные позиции

## Описание процесса очистки
Процесс предобработки сключает несколько этапов:
- Унификация дилеров и брендов: Исправлены неявные дубликаты. Текстовые вариации, такие как «Меркур Авто» и «Mercur Autos», приведены к единому стандарту Mercur Auto. 
- Временная конкатенация: Названия месяцев переведены в числовой формат и объединены со столбцами года. Сформирован единый признак «Дата продажи» в типе datetime64, где штамп времени установлен на последний день месяца, после чего избыточные столбцы были удалены.
- Нормализация географии и геополитических изменений: Строки регионов приведены к правильному заглавному регистру. Учтена реформа 2018 года: Южно-Казахстанская область переименована в Туркестанскую, а город Шымкент корректно извлечен из состава области и выделен в самостоятельный мегаполис республиканского значения.
- Стандартизация технических характеристик: Страны производства закодированы в международный стандарт ISO ALPHA-3 (KAZ, RUS, DEU). Для вида топлива данные закодированы в лаконичные категории (F, D, E, HYB). Приводы унифицированы до стандартов FWD/4WD/RWD. Весь массив трансмиссий из 88 хаотичных наименований (включая опечатки вроде 'CVT(вAриATор)') был строго разделен на две базовые категории: «Автомат» и «Механика». 
- Исправление числовых аномалий и пропусков: В столбце количества сохранены отрицательные значения (-1), идентифицированные как возвраты, а финансовый доход по ним скорректирован в минус для честного расчета итоговой выручки. Строки с критическими пропусками и техническими нулями в объемах двигателей были удалены для сохранения чистоты выборки. Оставшиеся значения округлены до одного знака после запятой, при этом точечно исправлены системные ошибки (например, сдвиг знака у Chevrolet Niva с 21.6 на 1.7 литра). 

## Ключевые выводы анализа рынка и позиций автоцентра «Меркур Авто»
- Общая картина рынка: Структура продаж в Казахстане за 2019 год ориентирована на розничный сегмент (более 75% сделок — это штучные продажи физическим лицам). Ядро масс-маркета формируют автомобили со средним чеком до 20 700 и объемом двигателя до 2.0 литров. При этом на рынке присутствуют крупные оптовые сделки (до 115 машин), которые генерируют разовые мощные всплески выручки.
- Позиционирование «Меркур Авто»: Автоцентр демонстрирует ярко выраженную премиальную стратегию. Холдинг удерживает монополию на бренды Volkswagen Group (Volkswagen, Audi, Porsche).
- Сильные стороны: Аномально высокий средний чек по сравнению с общерыночным за счет доминирования дорогих машин. Это обеспечивает колоссальную выручку при относительно небольших объемах продаж. Дополнительным фактором стабильности является база лояльных клиентов в крупнейших мегаполисах (Алматы и Нур-Султан) и высокий потенциал заработка на последующем сервисном обслуживании (запчасти и обслуживание).
- Слабые стороны и риски: Практически полное отсутствие холдинга в самом массовом и быстрорастущем сегменте ультрабюджетных автомобилей (до 15-20 тыс. долларов), где доминируют конкуренты. Бизнес критически зависит от квот и логистики одного немецкого концерна и слабо представлен в отдаленных регионах страны.
- Точки роста: Вектором развития для «Меркур Авто» может стать запуск и продвижение мультибрендовой программы Trade-In для подержанных автомобилей масс-маркета. Принимая в обмен бюджетные корейские и японские авто, холдинг сможет сформировать параллельный высокодоходный поток продаж, не размывая при этом премиальный статус своих основных автоцентров.

## Глубокая очистка и преобразование данных
На предыдущем этапе было выявлено много мусора, некорретных типов данных, ошибок - обработаем эти находки

```python
# Создадим копию датасета
data_clean = data.copy()
```

```python
# Преобразуем сразу дробные числа
pd.set_option('display.float_format', lambda x: '%.2f' % x)
```

```python
# Сформируем корректную дату - склеим год и месяц
# Создадим словарь для перевода названий месяцев в числа
data_clean['month'] = data_clean['month'].astype(str).str.strip().str.lower()
months_ru_to_num = {
    'январь': '01', 'февраль': '02', 'март': '03', 'апрель': '04',
    'май': '05', 'июнь': '06', 'июль': '07', 'август': '08',
    'сентябрь': '09', 'октябрь': '10', 'ноябрь': '11', 'декабрь': '12'
}

# Переводим текстовый месяц в строковое число
data_clean['month_cleaned'] = data_clean['month'].astype(str).str.strip().str.lower()
data_clean['month_num'] = data_clean['month_cleaned'].map(months_ru_to_num)

# Конкатенируем: Год + Месяц + Число '01' (добавили 1-е, т.к. оно есть во всех месяцах)
data_clean['date_str'] = (
    data_clean['year'].astype(str) + '-' + 
    data_clean['month_num'] + '-01'
)

# Превращаем строку в тип данных datetime
data_clean['date'] = pd.to_datetime(data_clean['date_str'], errors='coerce')

# Переносим дату на последний день месяца
data_clean['date'] = data_clean['date'] + pd.offsets.MonthEnd(0)

# Удаляем временные столбцы
data_clean = data_clean.drop(columns=['month', 'year', 'month_cleaned', 'month_num', 'date_str'])

print(data_clean[['date']].head())
```

```python
# Обработка названий компаний
data_clean['dealer_group'] = data_clean['dealer_group'].astype(str).str.strip()

dealer_mapping = {
    'Mercur Autos': 'Mercur Auto',
    'Каспиан Моторс': 'Caspian Motors',
    'Автокапитал': 'Autokapital',
    'ММС Рус': 'MMC RUS',
    'Равон Моторс Казахстан': 'Ravon Motors Kazakstan',
    'Хино Моторс Казахстан': 'Hino Motors',
    'Astana Motors ': 'Astana Motors',
    'Автомир-Центр': 'Автомир ГК',
    'ТОО "Eurasia Motor Zhaik"': 'Eurasia Motor',
    'Eurasia Motor Premium': 'Eurasia Motor'
}
data_clean['dealer_group'] = data_clean['dealer_group'].replace(dealer_mapping)
```

```python
# Обработка брендов
data_clean['brand'] = data_clean['brand'].astype(str).str.strip()

brand_mapping = {
    'Hyundai Truck & Bus': 'Hyundai',
    'Mercedes-Benz Trucks': 'Mercedes-Benz',
    'Nefaz': 'KAMAZ',
    'Daewoo': 'Ravon' 
}
data_clean['brand'] = data_clean['brand'].replace(brand_mapping)
```

```python
# Обработка стран
data_clean['country'] = data_clean['country'].astype(str).str.strip()

country_mapping = {
    'Германия': 'DEU',
        'США': 'USA',
        'Австрия': 'AUT',
        'Республика Казахстан': 'KAZ',
        'Российская Федерация': 'RUS',
        'Корея': 'KOR',
        'Япония': 'JPN',
        'Таиланд': 'THA',
        'Китай': 'CHN',
        'UK': 'GBR',
        'Узбекистан': 'UZB',
        'Венгрия': 'HUN',
        'Турция': 'TUR',
        'Испания': 'ESP',
        'Нидерланды': 'NLD',
        'Польша': 'POL',
        'Швеция': 'SWE',
        'Белоруссия': 'BLR',
        'Бельгия': 'BEL'
}

data_clean['country'] = data_clean['country'].map(country_mapping).fillna(data_clean['country'])
```

```python
# Обработка Вида топлива
data_clean['fuel_type'] = data_clean['fuel_type'].astype(str).str.strip().str.lower()

fuel_mapping = {
    'бензин': 'F',
    'дизель': 'D',
    'электро': 'E', 'электричество': 'E',
    'гибрид': 'HYB'
}
data_clean['fuel_type'] = data_clean['fuel_type'].map(fuel_mapping).fillna('UNKNOWN')
```

## Анализ рынка и позиции «Меркур Авто»

```python
# Создадим флаг, отделяющий "Меркур Авто" от остального рынка
data_clean['is_mercur'] = data_clean['dealer_group'] == 'Mercur Auto'
```

```python
# Считаем общие объемы для расчёта долей рынка
total_market_qty = data_clean['quantity'].sum()
total_mercur_qty = data_clean[data_clean['is_mercur']]['quantity'].sum()
mercur_market_share = (total_mercur_qty / total_market_qty) * 100

print(f"Общая доля Меркур Авто: {mercur_market_share:.2f}%")
```

Это весьма небольшой процент, который сигнализирует о потребности выделить сильные и слабые стороны, а также сосредоточиться на развитии потенциала компании

```python
# Анализ в разрезе брендов
brand_analysis = data_clean.groupby(['brand', 'is_mercur'])['quantity'].sum().unstack(fill_value=0)
brand_analysis.columns = ['Рынок_шт', 'Меркур_шт']
brand_analysis['Доля_Меркура_в_бренде_%'] = (brand_analysis['Меркур_шт'] / (brand_analysis['Рынок_шт'] + brand_analysis['Меркур_шт'])) * 100
brand_analysis['Продажи_Меркур_%'] = (brand_analysis['Меркур_шт'] / total_mercur_qty) * 100

print(brand_analysis[brand_analysis['Меркур_шт'] > 0].sort_values(by='Меркур_шт', ascending=False))
```

Меркур сосредоточен на продажах машин брендов Volkswagen (73.87%), Porsche (13.87%) и Audi (12.27%). Всего за период было продано 375 машин. И Меркур является монополистом на рынке по сбыту  Porsche и Audi, также высокий процент продаж в отношении конкурентов по моделям Volkswagen. 

Возможно, стоит расширить свой ассортимент и наладить поставки более популярных и продаваемых на рынке машин. Сейчас очевиден упор на продажи брендовых качественных машин - это формирует определенный статус компании, но может отпугивать среднестаистического покупателя, который ориентирован не на качество, а именно на соотношение цена+качество, где цена является нередко определяющим фактором

Данный график демонстрирует, что «Меркур Авто» осуществляет продажу не только люксовых моделей, но и представлен в масс-сегменте. Рассмотрим подробнее, какие авто продаются в среднем чеке

```python
# Выявим продаваемые в масс-сегменте марки "Меркур Авто"
mercur_data = data_clean[data_clean['is_mercur'] == True]

# Группируем по моделям, считаем сумму проданных штук и находим среднюю цену
top_mercur_models = mercur_data.groupby('model', as_index=False).agg(
    Продано_шт=('quantity', 'sum'),
    Средняя_цена=('price', 'mean')
)

# Определим топ-5
top_mercur_models = top_mercur_models.sort_values(by='Продано_шт', ascending=False).head(5).reset_index(drop=True)
top_mercur_models
```

Если ориенттироваться на средний чек, то среди наиболее продаваемых "Меркур" моделей обнаружилось весомое количество достаточно недорогих - было продуано 187 Polo со средней стоимостью - не выше 15 000$. Что заставляет нас пересмотреть картину

Автоцентры Меркура представлены в разных регионах. Стоит отметить, что в густонаселенных Алматы и Нур-Султане также есть центры продаж. Наиболее высокая доля продаж в Алматы, но относительно всего рынка показатели достаточно скромные - 4,64%. А в других регионах не переваливает за 2%

В целом, распределение по сегментам выглядит схожим, что по рынку - что у Меркура. Незначительное отставание Меркура обнаружено по продажам внедорожников и пикапов. По легковым машинам есть некоторое преимущество: 57.87% против 44.08%. Коммерческие автомобили и минивэны у Меркура продаются чуть хуже, чем по рынку. Нельзя сказать, что присутствуют аномалии, также стратегия продаж выглядит приемлемой - внедорожники и легковушки осатются самыми продаваемыми

Средний чек Меркура оказался выше на 11,5 тысяч чека по остальному рынку. Такой высокий средний чек сигнализирует, что холдинг продает меньше единиц товара, но зарабатывает на высокой маржинальности каждой сделки

```python
# Посчитаем динамику продаж для групп: Меркур vs Остальной рынок
monthly_comparison = data_clean.groupby(['date', 'Группа'])['quantity'].sum().reset_index()

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=monthly_comparison, 
    x='date', 
    y='quantity', 
    hue='Группа',
    marker='o', 
    linewidth=2.5,
    palette={'Меркур Авто': '#e65c00', 'Остальной рынок': '#2b5c8f'}
)

plt.title('Сравнение динамики продаж: Меркур Авто vs Остальной рынок', fontsize=14, pad=15)
plt.xlabel('Дата продажи')
plt.ylabel('Продано автомобилей (шт.)')
plt.xticks(rotation=15)
plt.legend(title='Группа')
plt.tight_layout()
```

На общем фоне выглядит так, что продажи Меркура относительно ситуации на рынке слишком незначительны. Также продажи Меркура на графике достаточно стабильны на протяжении всего периода (возможно, из-за масштаба незаметны резкие взлеты и падения на графике). Поэтому необходима детализированная информация по дилерам

```python
# Группируем данные по дилерам
dealer_rating = data_clean.groupby('dealer_group', as_index=False).agg(
    Продано_шт=('quantity', 'sum'),
    Выручка_всего=('income', 'sum')
)

# Считаем доли рынка для каждого дилера
total_qty = data_clean['quantity'].sum()
total_inc = data_clean['income'].sum()

dealer_rating['Доля в штуках'] = (dealer_rating['Продано_шт'] / total_qty) * 100
dealer_rating['Доля в выручке'] = (dealer_rating['Выручка_всего'] / total_inc) * 100

# Лидеры по количеству
top_dealers = dealer_rating.sort_values(by='Продано_шт', ascending=False).head(10)

# Сгруппируем столбики для графика
df_melted = pd.melt(
    top_dealers, 
    id_vars=['dealer_group'], 
    value_vars=['Доля в штуках', 'Доля в выручке'],
    var_name='Метрика рынка', 
    value_name='Процент от рынка (%)'
)

# Построим график
plt.figure(figsize=(12, 8))
sns.barplot(
    data=df_melted, 
    y='dealer_group', 
    x='Процент от рынка (%)', 
    hue='Метрика рынка', 
    palette={'Доля в штуках': '#2b5c8f', 'Доля в выручке': '#e65c00'}
)

plt.title('Рейтинг дилеров Казахстана: Доля в продажах (шт) vs Доля в выручке (деньги)', fontsize=14, pad=15)
plt.ylabel('Автодилер / Холдинг')
plt.xlabel('Доля от общего рынка (%)')
plt.legend(title='Бизнес-метрика')
plt.tight_layout()
```

Данный график наглядно демонстрирует лидеров по продажам как в количественном, так и финансовом отношениях. 8-е место из 40 дилеров занимает Меркур, что весьма неплохой результат. Но отставание от лидеров колоссальное. Необходимо детальнее ознакомиться со стратегиями лидеров, чтобы понять, что рекомендовать улучшать Меркур Авто

```python
# Сравним данные по трем интересующим нас компаниям
target_dealers = ['Mercur Auto', 'БИПЭК АВТО', 'Toyota Motor Kazakhstan']
df_leaders = data_clean[data_clean['dealer_group'].isin(target_dealers)].copy()

# Убираем категориальный тип
df_leaders['dealer_group'] = df_leaders['dealer_group'].astype(str)
df_leaders['segment'] = df_leaders['segment'].astype(str)

# Посчитаем количество проданных машин по сегментам
segment_shares = df_leaders.groupby(['dealer_group', 'segment'])['quantity'].sum().reset_index()

# Построим график
plt.figure(figsize=(14, 7))
sns.barplot(
    data=segment_shares,
    x='segment',
    y='quantity',
    hue='dealer_group',
    palette={'Mercur Auto': '#e65c00', 'БИПЭК АВТО': '#2b5c8f', 'Toyota Motor Kazakhstan': '#a6a6a6'}
)

plt.title('Сравнение структуры продаж по сегментам', fontsize=14, pad=15)
plt.xlabel('Маркетинговый сегмент автомобиля')
plt.ylabel('Количество проданных автомобилей (шт.)')
plt.xticks(rotation=15)
plt.legend(title='Дилер')
plt.tight_layout()
```

БИПЭК занимается продажами практически всех сегментов, за исключением пикапов, и удерживает лидирующие продажи. С такими крупными игроками Меркуру трудно тягаться, ведь продажи сосредоточены в основном в рамках легковых автомобилей. Однако, как раз в сегменте легковых автомобилей БИПЭК находится в глубоком отрыве, т.к. осуществляет продажу более бюджетных моделей, причем как выручка, так и количество продаж очень высокое. Toyota Motor Kazakhstan достаточно уверенно осуществляет продажи внедорожников, и в целом, Toyota пользовалась популярностью в Казахстане.
